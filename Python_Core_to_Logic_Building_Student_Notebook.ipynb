# 26. Recursion

**Recursion** occurs when a function calls itself.

Every useful recursive solution needs:
1. **Base case** — when recursion stops.
2. **Recursive case** — reduces the problem toward the base case.

Example: factorial

`5! = 5 × 4 × 3 × 2 × 1`

````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

# 36. Problem-Solving Example: Sum of Digits

### Problem
For `1234`, calculate `1 + 2 + 3 + 4 = 10`.

### Key idea
For a positive integer:
- `n % 10` extracts the last digit.
- `n // 10` removes the last digit.

### Trace

```text
1234 -> digit 4, remaining 123
123  -> digit 3, remaining 12
12   -> digit 2, remaining 1
1    -> digit 1, remaining 0
```

This is a foundational number-logic pattern.

````python
number = int(input("Enter a positive integer: "))

total = 0
n = number

while n > 0:
    digit = n % 10
    total += digit
    n //= 10

print("Sum of digits:", total)
````

# 37. Problem-Solving Example: Reverse a Number

Use the same digit-extraction pattern.

For `1234`:

```text
last digit = 4
reverse = 4

last digit = 3
reverse = 43

last digit = 2
reverse = 432

last digit = 1
reverse = 4321
```

General update:

```python
reverse = reverse * 10 + digit
```

````python
number = int(input("Enter a positive integer: "))

n = number
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

print("Reversed:", reverse)
````

# 38. Prime Number Logic

A **prime number** is an integer greater than 1 with exactly two positive divisors: `1` and itself.

### Beginner algorithm

For a number `n`:
- reject numbers below 2
- test possible divisors
- if any divisor divides `n` exactly, it is not prime

### Important optimization

You only need to test divisors up to `sqrt(n)`.

Why? If `n = a × b`, at least one of `a` or `b` must be at most `sqrt(n)`.

````python
def is_prime(n):
    if n < 2:
        return False

    divisor = 2
    while divisor * divisor <= n:
        if n % divisor == 0:
            return False
        divisor += 1

    return True

for number in range(1, 21):
    if is_prime(number):
        print(number, end=" ")
````

# 41. Frequency Counting

A very common logic pattern is:

> "How many times does each item occur?"

Use a dictionary as a frequency table.

Algorithm:
1. Start with an empty dictionary.
2. Visit every item.
3. Increase its count.
4. Return the frequency table.

````python
text = "banana"

frequency = {}

for character in text:
    frequency[character] = frequency.get(character, 0) + 1

print(frequency)
````

# 42. Searching

### Linear search

Check elements one by one until the target is found.

Time complexity: **O(n)** in the worst case.

This is an important first algorithm because it teaches:
- iteration
- comparison
- early exit
- Boolean results

````python
def linear_search(items, target):
    for index, item in enumerate(items):
        if item == target:
            return index
    return -1

numbers = [10, 25, 7, 42, 19]

print(linear_search(numbers, 42))
print(linear_search(numbers, 100))
````

# 43. Edge Cases

An **edge case** is an input at or near a boundary where program behavior may be different.

For a function working with numbers, consider:
- `0`
- `1`
- negative values
- very large values

For lists:
- empty list
- one element
- duplicates
- already sorted data

Good programmers actively search for edge cases.

````python
def first_item(items):
    if not items:
        return None
    return items[0]

print(first_item([]))
print(first_item([42]))
````

# 44. Time Complexity — Beginner Introduction

**Time complexity** describes how the amount of work grows as input size grows.

Common patterns:

- One loop over `n` items → often **O(n)**
- Two nested loops over `n` items → often **O(n²)**
- Repeatedly halving the search space → often **O(log n)**

Do not memorize blindly. Count how many times the important operation executes as input size grows.

````python
# Roughly O(n): one pass
numbers = list(range(100))

total = 0
for number in numbers:
    total += number

print(total)
````
