## 13. List, Dict, and Set Comprehensions

````python
nums = range(1, 11)

# List comprehension
squares = [n * n for n in nums]
print(squares)

# With a filter
even_squares = [n * n for n in nums if n % 2 == 0]
print(even_squares)

# Dict comprehension
square_map = {n: n * n for n in range(1, 6)}
print(square_map)

# Set comprehension
lengths = {len(word) for word in ["hi", "hello", "hey", "howdy"]}
print(lengths)
````

## 16. Scope

- Names defined inside a function are local by default.
- `global` lets a function change a module-level name.
- `nonlocal` lets an inner function change an enclosing function's name.


````python
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
print("counter:", counter)


def make_counter():
    value = 0
    def step():
        nonlocal value
        value += 1
        return value
    return step

next_value = make_counter()
print(next_value(), next_value(), next_value())
````

## 20. Exceptions

An exception is an error signaled at runtime. Catch it with `try / except`. Use `raise` to signal your own errors. `else` runs when there was no exception. `finally` always runs.

````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero.")
        return None
    except TypeError as error:
        print("Type error:", error)
        return None
    else:
        print("Division succeeded.")
        return result
    finally:
        print("safe_divide finished.")

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "x"))


class NegativeAgeError(ValueError):
    """Raised when a supplied age is negative."""


def set_age(age):
    if age < 0:
        raise NegativeAgeError(f"Age cannot be negative: {age}")
    return age

try:
    set_age(-5)
except NegativeAgeError as error:
    print("Caught:", error)
````

## 25. Class Attributes, Class Methods, Static Methods

- **Instance attributes** live on the object (`self.x`).
- **Class attributes** live on the class and are shared by all instances.
- **Class methods** take the class itself as first argument (`cls`).
- **Static methods** do not take `self` or `cls`; they are just plain functions grouped with the class.


````python
class Circle:
    pi = 3.14159  # class attribute

    def __init__(self, radius):
        self.radius = radius  # instance attribute

    def area(self):
        return Circle.pi * self.radius ** 2

    @classmethod
    def unit_circle(cls):
        return cls(1)

    @staticmethod
    def describe():
        return "A round shape."

c = Circle(5)
print(c.area())
print(Circle.unit_circle().area())
print(Circle.describe())
````

## 26. Inheritance

A subclass can extend or override a parent class. Use `super()` to call the parent's version of a method.

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "Some generic sound"


class Cat(Animal):
    def __init__(self, name, color):
        super().__init__(name)
        self.color = color

    def speak(self):
        parent_sound = super().speak()
        return f"{self.name} (a {self.color} cat) says meow. Parent said: {parent_sound}"

kitty = Cat("Whiskers", "black")
print(kitty.speak())
````

## 27. Special (Dunder) Methods

Dunder methods let your objects work with Python syntax. For example, defining `__len__` makes `len(obj)` work.

````python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}"

    def __repr__(self):
        return f"Book(title={self.title!r}, pages={self.pages})"

    def __len__(self):
        return self.pages

    def __eq__(self, other):
        return isinstance(other, Book) and self.title == other.title

b1 = Book("Python 101", 320)
b2 = Book("Python 101", 350)

print(str(b1))
print(repr(b1))
print(len(b1))
print(b1 == b2)
````

## 28. Properties

`@property` turns a method into a read-only attribute. A matching setter can validate the value being assigned.

````python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero.")
        self._celsius = value

    @property
    def fahrenheit(self):
        return self._celsius * 9 / 5 + 32

t = Temperature(25)
print(t.celsius, t.fahrenheit)
t.celsius = 30
print(t.fahrenheit)
````

## 29. Dataclasses

## 30. Abstract Base Classes

An abstract class defines an interface. Subclasses must implement its abstract methods before they can be instantiated.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        ...

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

s = Square(4)
print(s.area())

try:
    Shape()
except TypeError as error:
    print("Cannot instantiate Shape directly:", error)
````

## 31. Iterators

Any object that implements `__iter__` and `__next__` is an iterator. `for` loops use this protocol under the hood. `next()` fetches the next value; a `StopIteration` exception signals the end.

````python
class Countdown:
    def __init__(self, start):
        self.value = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.value <= 0:
            raise StopIteration
        current = self.value
        self.value -= 1
        return current

for n in Countdown(3):
    print(n)
````

## 32. Generators

A function with `yield` returns a generator. Generators produce values lazily, one at a time, without building the full list in memory.

````python
def fibonacci(limit):
    a, b = 0, 1
    while a < limit:
        yield a
        a, b = b, a + b

print(list(fibonacci(50)))

# Generator expression: same shape as a list comprehension but with parentheses.
squares_gen = (n * n for n in range(5))
print(sum(squares_gen))
````

## 33. Decorators

A decorator wraps a function to add behavior. Use `functools.wraps` to keep the original function's name and docstring. Decorators can also take arguments.

````python
import functools
import time

def timed(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = time.perf_counter() - start
        print(f"{func.__name__} took {elapsed*1000:.2f} ms")
        return result
    return wrapper

@timed
def slow_square(n):
    """Square n after a tiny pause."""
    time.sleep(0.05)
    return n * n

print(slow_square(9))
print(slow_square.__name__, "-", slow_square.__doc__)


# Parameterized decorator
def repeat(times):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(times):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(times=3)
def hello():
    print("hi")

hello()
````

## 34. Context Managers

The `with` statement runs setup and cleanup around a block of code. You can write one as a class with `__enter__` and `__exit__`, or with the `contextlib.contextmanager` decorator.

````python
from contextlib import contextmanager

class Section:
    def __init__(self, name):
        self.name = name

    def __enter__(self):
        print(f"--- start {self.name} ---")
        return self

    def __exit__(self, exc_type, exc, tb):
        print(f"--- end {self.name} ---")

with Section("work"):
    print("doing work")


@contextmanager
def timer(label):
    import time
    start = time.perf_counter()
    yield
    print(f"{label} took {(time.perf_counter() - start)*1000:.2f} ms")

with timer("sum loop"):
    total = sum(range(100_000))
print("total:", total)
````

## 35. Type Hints

Type hints describe the expected types of arguments and return values. They are optional and do not enforce anything at runtime, but they make code easier to read and enable tools like `mypy` to catch bugs.

````python
from typing import Optional, Union, Callable

def add(a: int, b: int) -> int:
    return a + b

def find_user(user_id: int) -> Optional[dict]:
    if user_id == 1:
        return {"id": 1, "name": "Ada"}
    return None

def format_value(value: Union[int, float, str]) -> str:
    return f"value={value}"

# Modern (Python 3.9+) container syntax
def average(values: list[float]) -> float:
    return sum(values) / len(values)

# Callable
def apply(op: Callable[[int, int], int], a: int, b: int) -> int:
    return op(a, b)

print(add(2, 3))
print(find_user(1))
print(format_value(3.5))
print(average([1.0, 2.0, 3.0]))
print(apply(lambda x, y: x * y, 4, 5))
````

## 36. Concurrency Overview

Python offers several ways to do work "at the same time":

- **Threads** (`threading`): good for I/O-bound work like network calls.
- **Processes** (`multiprocessing`): good for CPU-bound work, because they sidestep the Global Interpreter Lock.
- **`concurrent.futures`**: a friendlier interface over both threads and processes.
- **`asyncio`**: a single-threaded cooperative model for I/O-heavy code (next section).

Below is a `ThreadPoolExecutor` example, which is safe and easy to run inside a notebook.

````python
from concurrent.futures import ThreadPoolExecutor
import time

def fetch(item):
    time.sleep(0.1)   # pretend this is a slow I/O call
    return f"fetched {item}"

items = ["A", "B", "C", "D"]

start = time.perf_counter()
with ThreadPoolExecutor(max_workers=4) as pool:
    results = list(pool.map(fetch, items))
elapsed = time.perf_counter() - start

print(results)
print(f"Elapsed: {elapsed:.2f} s (would be ~0.4 s if run sequentially)")
````

## 37. `asyncio` Basics

`async def` defines a coroutine. `await` pauses it until another coroutine finishes. Multiple coroutines can run concurrently in a single thread with `asyncio.gather`.

Note about notebooks: Jupyter already runs an event loop, so `await` works at the top level of a cell. In a plain script you would use `asyncio.run(main())` instead.

````python
import asyncio
import threading
import time

async def task(name, seconds):
    await asyncio.sleep(seconds)
    return f"{name} done after {seconds}s"

async def main():
    start = time.perf_counter()
    results = await asyncio.gather(
        task("A", 0.3),
        task("B", 0.2),
        task("C", 0.1),
    )
    elapsed = time.perf_counter() - start
    return results, elapsed

def run_async(coro_factory):
    """Run an async function whether or not an event loop is already running.

    Plain Python scripts have no running loop, so asyncio.run works directly.
    Jupyter already has a running loop, so we run the coroutine on a fresh
    loop in a background thread instead.
    """
    try:
        return asyncio.run(coro_factory())
    except RuntimeError:
        box = {}
        def worker():
            loop = asyncio.new_event_loop()
            try:
                box["value"] = loop.run_until_complete(coro_factory())
            finally:
                loop.close()
        thread = threading.Thread(target=worker)
        thread.start()
        thread.join()
        return box["value"]

results, elapsed = run_async(main)
print(results)
print(f"Elapsed: {elapsed:.2f} s (would be ~0.6 s if run sequentially)")
````

## 38. Logging

`print` is fine for quick debugging, but real programs use the `logging` module. It supports levels (`DEBUG`, `INFO`, `WARNING`, `ERROR`, `CRITICAL`), formatting, and multiple output destinations.

````python
import logging

logger = logging.getLogger("demo")
logger.setLevel(logging.DEBUG)

# Fresh handler so re-running this cell does not duplicate output.
for h in list(logger.handlers):
    logger.removeHandler(h)

handler = logging.StreamHandler()
handler.setFormatter(logging.Formatter("%(levelname)s: %(message)s"))
logger.addHandler(handler)

logger.debug("Debug message")
logger.info("Starting job")
logger.warning("Something looks off")
logger.error("Something failed")
````

## 39. Testing with `unittest`

`unittest` is Python's built-in test framework. Group tests in a class that inherits from `unittest.TestCase` and use `self.assert...` methods.

````python
import unittest

def add(a, b):
    return a + b

class AddTests(unittest.TestCase):
    def test_positive(self):
        self.assertEqual(add(2, 3), 5)

    def test_negative(self):
        self.assertEqual(add(-1, -1), -2)

    def test_zero(self):
        self.assertEqual(add(0, 5), 5)

# Run the tests inside the notebook.
unittest.main(argv=["ignored", "-v"], exit=False)
````

## 40. Performance Tips

- Use `timeit` to measure small snippets.
- Prefer built-ins and comprehensions over hand-written loops.
- Generators use very little memory compared to building large lists.
- For heavy numeric work, reach for NumPy or pandas (see the data-science notebooks in this folder).


````python
import timeit
import sys

setup = "data = list(range(10_000))"

t_loop = timeit.timeit("total = 0\nfor x in data: total += x*x", setup=setup, number=200)
t_comp = timeit.timeit("total = sum([x*x for x in data])", setup=setup, number=200)
t_gen  = timeit.timeit("total = sum(x*x for x in data)", setup=setup, number=200)

print(f"for loop:            {t_loop:.3f} s")
print(f"list comprehension:  {t_comp:.3f} s")
print(f"generator:           {t_gen:.3f} s")

# Memory: a list holds all values, a generator does not.
big_list = [x for x in range(100_000)]
big_gen = (x for x in range(100_000))
print("list size (bytes):     ", sys.getsizeof(big_list))
print("generator size (bytes):", sys.getsizeof(big_gen))
````
