## 6. Loops

`for` loops iterate over sequences (lists, strings, ranges). `while` loops repeat while a condition is true.

````python
# for loop with range
for i in range(0, 100, 10):
    if i == 60:
        continue
    print("hello")
    print(f"Count:{i}" )

print("---")

# for loop over a list
subjects_list = ["Python", "Django", "SQL", "GENAI", 100, {'Course': 'GenAI', 'Instructor': 'Hrithik'}]



for object in subjects_list:
    print(f"Learning: {object}")

print("---")


# enumerate - gives index + value (very common in real code)
for index, object in enumerate(subjects_list):
    print(f"{index} - {object}")

print("---")




# while loop
count = 0
while count < 3:
    print(f"While iteration: {count}")
    count += 1
````

### `break` and `continue`

````python
for i in range(1, 10):
    if i == 5:
        break          # stops the loop entirely
    if i % 2 == 0:
        continue        # skips this iteration
    print(i)
````

````python
# *args and **kwargs - variable number of arguments
def add_all(*numbers):
    addtiion = 0
    for num in numbers:
        addtiion += num
    return addtiion

def print_profile(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

print(add_all(1, 2, 3, 4, 5,))
print_profile(name="Ravi", age=22, city="Mumbai",address="Pune",course="Python")
````

## 9. Object-Oriented Programming (OOP)

Django is built **entirely on OOP** — Models, Views (class-based), and Forms are all Python classes. Understanding classes is non-negotiable before Django.

````python
class Student:
    college_name = "ABC Institute"   # class attribute (shared by all instances)

    def __init__(self, name, age):   # constructor
        self.name = name              # instance attribute
        self.age = age

    def introduce(self):              # instance method
        return f"Hi, I'm {self.name}, {self.age} years old."

s1 = Student("Meera", 20)
s2 = Student("Arjun", 22)

print(s1.introduce())
print(s2.introduce())
print(s1.college_name, s2.college_name)
````

### Inheritance
A class can inherit attributes/methods from another class. Django's `models.Model`, `forms.Form`, and class-based views all work through inheritance — this is exactly the pattern you'll see.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello, I am {self.name}"


class Teacher(Person):                 # Teacher inherits from Person
    def __init__(self, name, subject):
        super().__init__(name)          # call parent constructor
        self.subject = subject

    def greet(self):                    # method overriding
        base_greeting = super().greet()
        return f"{base_greeting} and I teach {self.subject}"


t = Teacher("Mr. Sharma", "Django")
print(t.greet())
print(isinstance(t, Person))    # True - Teacher IS-A Person
````

## 10. Exception Handling

Real applications must handle errors gracefully instead of crashing — Django relies on this constantly (e.g., handling "Object does not exist" errors).

````python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Invalid types provided!")
        return None
    else:
        print("Division successful!")
        return result
    finally:
        print("Divide operation attempted.\n")

print(divide(10, 2))
print(divide(10, 0))
print(divide(10, "a"))
````

## 13. Introduction to Django

### What is Django?
Django is a **high-level Python web framework** that enables rapid development of secure, maintainable websites. It's "batteries-included" — meaning it ships with almost everything you need: an ORM, admin panel, authentication, form handling, and more.

### Why Django?
- Used by Instagram, Spotify, Pinterest, YouTube (early days).
- Follows the **DRY** principle (Don't Repeat Yourself).
- Built-in **security** (protects against SQL injection, XSS, CSRF by default).
- Comes with a powerful **Admin Panel** for free.

### The MVT Architecture
Django follows **MVT (Model–View–Template)**, a variant of MVC:

| Component | Responsibility |
|---|---|
| **Model** | Defines the data structure (talks to the database) — Session 2 |
| **View** | Contains the business logic; takes a request, returns a response — *today* |
| **Template** | The HTML that gets rendered to the user — Session 2 |
| **URL dispatcher** | Maps a URL to the correct view function — *today* |

**Request flow:**
```
Browser Request → urls.py (finds matching URL) → views.py (processes logic) → template.html (renders output) → Response back to Browser
```

We will build the **URL → View** part today. Models & Templates come in Session 2.

## 15. Django Project Structure Explained

After running `django-admin startproject myschool`, you get:

```
myschool/
├── manage.py            <- Command-line utility to run/manage the project
└── myschool/
    ├── __init__.py       <- Marks this folder as a Python package
    ├── settings.py        <- ALL project configuration (database, apps, etc.)
    ├── urls.py             <- The root URL configuration ("table of contents")
    ├── asgi.py             <- Entry point for async servers (advanced)
    └── wsgi.py             <- Entry point for traditional servers (deployment)
```

### Key files explained:

| File | Purpose |
|---|---|
| `manage.py` | Run this to start the server, create apps, run migrations, etc. |
| `settings.py` | Configure installed apps, database, templates, static files, timezone |
| `urls.py` | Maps URLs (like `/about/`) to views |
| `wsgi.py` / `asgi.py` | Used when deploying to a real server (not touched by beginners) |

**Important terminology:**
- A **project** = the whole website/configuration.
- An **app** = a self-contained module inside the project (e.g., `blog`, `students`, `payments`). A project can have many apps.

## 17. URLs & Views — "Hello World"

A **view** is just a Python function that takes an `HttpRequest` and returns an `HttpResponse`.

### Step 1 — Write a view in `students/views.py`
```python
from django.http import HttpResponse

def home(request):
    return HttpResponse("Hello World! Welcome to Django.")

def about(request):
    return HttpResponse("This is the About page for the Students app.")
```

### Step 2 — Create `students/urls.py` (app-level routing)
```python
from django.urls import path
from . import views

urlpatterns = [
    path('', views.home, name='home'),
    path('about/', views.about, name='about'),
]
```

### Step 3 — Include the app's urls in the project's `myschool/urls.py`
```python
from django.contrib import admin
from django.urls import path, include

urlpatterns = [
    path('admin/', admin.site.urls),
    path('', include('students.urls')),
]
```


````python
views_code = '''from django.http import HttpResponse


def home(request):
    return HttpResponse("Hello World! Welcome to Django.")


def about(request):
    return HttpResponse("This is the About page for the Students app.")
'''

with open("students/views.py", "w") as f:
    f.write(views_code)

print("students/views.py written:\n")
print(views_code)
````

````python
app_urls_code = '''from django.urls import path
from . import views

urlpatterns = [
    path('', views.home, name='home'),
    path('about/', views.about, name='about'),
]
'''

with open("students/urls.py", "w") as f:
    f.write(app_urls_code)

print("students/urls.py written:\n")
print(app_urls_code)
````

````python
project_urls_path = "myschool/urls.py"

with open(project_urls_path, "r") as f:
    content = f.read()

content = content.replace(
    "from django.urls import path",
    "from django.urls import path, include"
)
content = content.replace(
    "path('admin/', admin.site.urls),",
    "path('admin/', admin.site.urls),\n    path('', include('students.urls')),"
)

with open(project_urls_path, "w") as f:
    f.write(content)

print("myschool/urls.py updated:\n")
!cat myschool/urls.py
````
