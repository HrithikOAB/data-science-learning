## 1. What is Programming? What is Python?

**Programming** is the process of writing a precise set of instructions that a computer can follow, in order to solve a problem or perform a task. A computer only understands binary (0s and 1s) — a **programming language** is a human-readable way to describe instructions that eventually get translated into something the machine can execute.

**Python** is a **high-level, general-purpose, interpreted programming language** created by Guido van Rossum, first released in 1991.

Let's break down every word in that definition, because each one matters:

| Term | Meaning |
|---|---|
| **High-level** | Close to human language, far from machine code. You write `x = a + b`, not raw binary/CPU instructions. |
| **General-purpose** | Not built for one narrow task — used for web development, data science, automation, AI/ML, scripting, and more. |
| **Interpreted** | Code is read and executed line-by-line by a program called an **interpreter**, rather than being fully translated to machine code ahead of time (as in **compiled** languages like C++). |

### Why Python is popular
- **Readable syntax** — often reads close to plain English.
- **Huge ecosystem** — libraries for almost anything (`django` for web apps, `pandas` for data, `numpy` for math, etc.).
- **Beginner-friendly**, yet powerful enough for large, serious systems (Instagram, YouTube, Spotify backend services all use Python).


## 2. How Python Runs Your Code

When you run a `.py` file:
1. The **Python interpreter** reads your source code.
2. It converts it into an intermediate form called **bytecode**.
3. The **Python Virtual Machine (PVM)** executes that bytecode line by line.

This is different from a language like C, where a **compiler** translates the *entire* program into machine code **before** it ever runs. Python's line-by-line interpreted nature is why:
- You get errors only when the interpreter *reaches* the broken line (not necessarily at the very start).
- Python is generally slower than compiled languages, but much faster to write and debug.

### 3.3 Indentation — Python's Most Distinctive Rule
Most languages use `{ }` (curly braces) to group blocks of code. **Python uses indentation (whitespace) instead.** This is not a style choice — it is **required syntax**. Code at the same indentation level is understood to belong to the same block.

## 4. Variables

### Definition
A **variable** is a named reference to a value stored in the computer's memory. Think of it as a labeled box: the label is the variable name, and the box holds a value you can retrieve, use, or change later.

In Python, you create a variable simply by assigning a value to a name using `=` (the **assignment operator**):
```python
variable_name = value
```

### Rules for naming a variable
- Must start with a letter (a-z, A-Z) or an underscore `_` — **never** a digit.
- Can contain letters, digits, and underscores after the first character.
- **Case-sensitive**: `age`, `Age`, and `AGE` are three different variables.
- Cannot be a **reserved keyword** (words Python has already claimed, like `if`, `for`, `class`, `return`).

### Python is dynamically typed
You never declare a variable's type in advance — Python figures it out automatically from the value assigned, and a variable can even be reassigned to a completely different type later. (Compare this to languages like Java or C, where you must write `int age = 21;` and `age` can *only* ever hold an integer.)


````
# Creating variables
student_name = "Aditi"
age = 21
height_in_cm = 165.5

print(student_name)
print(age)
print(height_in_cm)

# Dynamic typing in action - the SAME variable can hold different types over time
value = 10
print(value, type(value))

value = "now I am text"
print(value, type(value))
````

````
# Multiple assignment - Python-specific convenience syntax
x, y, z = 1, 2, 3
print(x, y, z)

# Assigning the same value to multiple variables at once
a = b = c = 0
print(a, b, c)
````

### Valid vs Invalid Variable Names

````
# Valid names
_score = 10
score2 = 20
total_marks = 30
totalMarks = 40   # valid, though snake_case (total_marks) is Python's preferred convention

print("All valid names work fine.")

# Invalid names would cause a SyntaxError if uncommented:
# 2score = 10        <- cannot start with a digit
# total-marks = 20   <- hyphens are not allowed
# class = "A"        <- 'class' is a reserved keyword
````

## 5. Data Types — Deep Dive

### Definition
A **data type** defines what *kind* of value a variable holds, and consequently what operations can be performed on it. Python has several **built-in data types**, grouped as follows:

| Category | Types |
|---|---|
| Numeric | `int`, `float`, `complex` |
| Text | `str` |
| Boolean | `bool` |
| Sequence | `list`, `tuple`, `range` |
| Mapping | `dict` |
| Set | `set`, `frozenset` |
| None | `NoneType` |

We'll cover the numeric/text/boolean types deeply here, and sequences/mappings/sets in Section 15.

Use the built-in function **`type(x)`** at any point to check the exact data type of a value `x`.


````
print(type(10))          # int
print(type(10.5))         # float
print(type(3 + 4j))       # complex - has a real and imaginary part
print(type("hello"))      # str
print(type(True))         # bool
print(type(None))         # NoneType
````

### 5.1 Numeric Types
- **`int`** — a whole number, positive or negative, with no size limit in Python (unlike many other languages).
- **`float`** — a number with a decimal point, used to represent real numbers with fractional parts.
- **`complex`** — a number with a real and an imaginary part (`a + bj`), used in scientific/mathematical computing. You will rarely need this as a beginner.


````
whole_number = 42
decimal_number = 3.14159
negative_number = -7

print(whole_number, type(whole_number))
print(decimal_number, type(decimal_number))
print(negative_number, type(negative_number))

# Python integers have NO fixed size limit (unlike C's int, which overflows)
huge_number = 2 ** 100
print(huge_number)
````

### 5.2 Text Type: `str` (String)
A **string** is an ordered sequence of characters, used to represent text. Strings are written inside single quotes `'...'`, double quotes `"..."`, or triple quotes `'''...'''` / `"""..."""` (the latter used for multi-line strings).

Strings are **immutable** — meaning once a string is created, its individual characters cannot be changed in place. (We'll see exactly what this means in Section 9.)


````
single = 'Hello'
double = "World"
multiline = """This string
spans multiple
lines."""

print(single, double)
print(multiline)
````

### 5.3 Boolean Type: `bool`
A **boolean** represents one of exactly two values: `True` or `False`. Booleans are the foundation of all decision-making in code (Section 10) — they are typically the *result* of a comparison.

Internally, `True` behaves like `1` and `False` behaves like `0` — you can even do arithmetic with them, though this is rarely done in practice.


````
is_raining = True
is_sunny = False

print(is_raining, type(is_raining))

# Booleans are the result of comparisons
print(5 > 3)          # True
print(5 == 3)          # False

# Booleans behave like integers under the hood
print(True + True)      # 2
print(False + 5)         # 5
````

### 5.4 `NoneType`
`None` is a special, singular value in Python representing the **absence of a value** — "nothing here," not zero, not an empty string, not `False`. It is its own type, `NoneType`, and there is only ever one `None`.

You'll commonly see `None` used as:
- A default/placeholder value before real data is available.
- The value a function returns when it doesn't explicitly `return` anything (Section 13).


````
result = None
print(result, type(result))

print(result == None)      # True
print(result is None)      # True - "is" checks identity, the preferred way to check for None
````

### 5.5 Mutable vs Immutable — A Core Concept
- **Immutable**: the value cannot be changed after creation. Any "modification" actually creates a brand-new object. Examples: `int`, `float`, `bool`, `str`, `tuple`.
- **Mutable**: the value CAN be changed in place, without creating a new object. Examples: `list`, `dict`, `set`.

This distinction matters constantly in real programs (e.g., why strings can't be edited character-by-character, why passing a list into a function can let that function change your original data). We'll revisit this concretely in Sections 9 and 15.


````
# Immutability of integers - reassignment creates a NEW object, doesn't change the old one
a = 10
print(id(a))     # id() shows a unique identifier for the object in memory
a = a + 1
print(id(a))     # different id - 'a' now points to a DIFFERENT object entirely

# Mutability of lists - the SAME object is modified in place
my_list = [1, 2, 3]
print(id(my_list))
my_list.append(4)
print(id(my_list))   # SAME id - the original object was modified, not replaced
````

## 6. Type Conversion (Casting)

**Type conversion** is changing a value from one data type to another.

- **Implicit conversion**: Python does it automatically, when it's "safe" (no data is lost).
- **Explicit conversion (casting)**: You do it deliberately, using functions like `int()`, `float()`, `str()`, `bool()`.


````
# Implicit conversion - Python automatically converts int to float here so the math makes sense
x = 5          # int
y = 2.5        # float
result = x + y
print(result, type(result))   # int + float = float, automatically
````

````
# Explicit conversion (casting)
age_text = "25"
age_number = int(age_text)         # str -> int
print(age_number + 5)

pi_text = "3.14"
pi_number = float(pi_text)          # str -> float
print(pi_number * 2)

score = 90
score_text = str(score)              # int -> str
print("Score: " + score_text)

# Casting to bool: nearly everything is True EXCEPT these "falsy" values:
# 0, 0.0, "", [], {}, (), None, False
print(bool(0))          # False
print(bool(""))          # False
print(bool("hi"))         # True
print(bool([]))            # False
print(bool([1, 2]))         # True
````

````
# A conversion that FAILS - not every string can become a number
try:
    int("hello")
except ValueError as e:
    print("Conversion failed:", e)
````

## 7. Input & Output

### `input()`
`input(prompt)` pauses the program, displays `prompt` to the user, waits for them to type something and press Enter, and returns **whatever they typed, always as a string** — even if they type a number.

### `print()` — deeper look
`print()` accepts multiple values and two especially useful keyword arguments:
- `sep` — what to put *between* multiple printed values (default is a space).
- `end` — what to put *after* the entire print call (default is a newline `\n`).


````
name = input("Enter your name: ")
print("Hello,", name)
print(type(name))   # always <class 'str'>, even if the user typed a number
````

````
# print() with sep and end
print("a", "b", "c")                      # default sep=' '
print("a", "b", "c", sep="-")              # custom separator
print("Loading", end="...")                 # no newline after
print("Done!")                                # continues on the SAME line as above

print(1, 2, 3, sep=", ", end=".\n")           # combining both
````

## 8. Operators — Every Category

An **operator** is a symbol that performs an operation on one or more values (called **operands**).

### 8.1 Arithmetic Operators
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | Addition | `7 + 3` | `10` |
| `-` | Subtraction | `7 - 3` | `4` |
| `*` | Multiplication | `7 * 3` | `21` |
| `/` | Division (always returns float) | `7 / 2` | `3.5` |
| `//` | Floor division (drops the decimal part) | `7 // 2` | `3` |
| `%` | Modulus (remainder) | `7 % 2` | `1` |
| `**` | Exponentiation (power) | `2 ** 3` | `8` |


````
a, b = 17, 5
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("True Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
````

### 8.2 Comparison (Relational) Operators
These always produce a **boolean** result (`True`/`False`).

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |


````
x, y = 10, 20
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= 10)
print(y <= 20)
````

### 8.3 Logical Operators
Combine multiple boolean expressions.

| Operator | Meaning |
|---|---|
| `and` | `True` only if **both** sides are `True` |
| `or` | `True` if **at least one** side is `True` |
| `not` | Flips `True` to `False` and vice versa |


````
age = 25
has_id = True

print(age >= 18 and has_id)     # both must be True
print(age < 18 or has_id)         # at least one True
print(not has_id)                  # flips True -> False
````

### 8.4 Assignment Operators
Shorthand for updating a variable based on its own current value.

| Operator | Equivalent To |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |
| `x //= 5` | `x = x // 5` |
| `x %= 5` | `x = x % 5` |
| `x **= 5` | `x = x ** 5` |


````
counter = 10
counter += 5    # same as counter = counter + 5
print(counter)

counter *= 2
print(counter)

counter -= 10
print(counter)
````

### 8.5 Membership Operators
Check whether a value exists inside a sequence (string, list, tuple, etc).

| Operator | Meaning |
|---|---|
| `in` | `True` if the value IS found |
| `not in` | `True` if the value is NOT found |


````
fruits = ["apple", "banana", "mango"]
print("apple" in fruits)
print("grape" not in fruits)
print("a" in "banana")   # works on strings too - checks for substring
````

### 8.6 Identity Operators
Check whether two variables refer to the **exact same object in memory** (not just equal values).

| Operator | Meaning |
|---|---|
| `is` | `True` if same object |
| `is not` | `True` if different objects |


````
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)     # True - same VALUES
print(a is b)      # False - different OBJECTS in memory, even though values match
print(a is c)       # True - c points to the exact same object as a
````

### 8.7 Bitwise Operators
Operate on the individual **binary bits** of integers. Less common for beginners, but good to know they exist.

| Operator | Meaning |
|---|---|
| `&` | Bitwise AND |
| `\|` | Bitwise OR |
| `^` | Bitwise XOR |
| `~` | Bitwise NOT |
| `<<` | Left shift |
| `>>` | Right shift |


````
a, b = 6, 3   # in binary: 6 = 110, 3 = 011
print(a & b)    # 010 = 2
print(a | b)     # 111 = 7
print(a ^ b)      # 101 = 5
print(a << 1)      # 1100 = 12 (shift left = multiply by 2)
print(a >> 1)        # 011 = 3 (shift right = divide by 2)
````

### 8.8 Operator Precedence
When an expression has multiple operators, Python evaluates them in a specific order (similar to mathematical order of operations — PEMDAS). From **highest to lowest** priority (simplified):

```
**                (exponent)
* / // %          (multiplication, division, floor division, modulus)
+ -                (addition, subtraction)
== != > < >= <=      (comparisons)
not
and
or
```

Use **parentheses `()`** whenever you want to be 100% explicit and avoid ambiguity — it costs nothing and prevents subtle bugs.


````
result = 2 + 3 * 4          # multiplication happens first: 2 + 12
print(result)                 # 14

result2 = (2 + 3) * 4          # parentheses force addition first
print(result2)                  # 20

result3 = 10 > 5 and 3 < 5        # comparisons happen before 'and'
print(result3)                     # True
````

## 9. Strings — In Depth

We defined strings in Section 5. Now let's go deep, since strings are used constantly in real programs.

### 9.1 Indexing
Every character in a string has a **position (index)**, starting from `0` for the first character. Python also supports **negative indexing**, where `-1` refers to the last character, `-2` the second-last, and so on.


````
word = "PYTHON"
#        012345    (positive indices)
#       -6-5-4-3-2-1  (negative indices)

print(word[0])     # 'P' - first character
print(word[5])      # 'N' - last character by positive index
print(word[-1])       # 'N' - last character by negative index
print(word[-6])        # 'P' - first character by negative index
````

### 9.2 Slicing
**Slicing** extracts a sub-portion of a string using the syntax `string[start:stop:step]`:
- `start` — index to begin at (inclusive). Default: beginning.
- `stop` — index to stop before (**exclusive** — this index is NOT included). Default: end.
- `step` — how many characters to skip each time. Default: `1`.


````
word = "PROGRAMMING"

print(word[0:4])      # 'PROG' - characters at index 0,1,2,3 (4 is excluded)
print(word[:4])         # same as above - start defaults to 0
print(word[4:])           # 'RAMMING' - from index 4 to the end
print(word[:])              # 'PROGRAMMING' - the whole string (a full copy)
print(word[::2])              # every 2nd character: 'PORMIG'
print(word[::-1])               # step -1 reverses the string entirely: 'GNIMMARGORP'
````

### 9.3 Strings Are Immutable — Proven
As stated in Section 5, you **cannot** change a character of a string in place.


````
word = "Hello"
try:
    word[0] = "J"     # attempting to change just the first character
except TypeError as e:
    print("Error:", e)

# The correct way is to build a NEW string instead
word = "J" + word[1:]
print(word)   # "Jello" - this is a brand new string object, not a modified original
````

### 9.4 String Concatenation and Repetition


````
first = "Py"
second = "thon"
print(first + second)     # concatenation
print(first * 3)            # repetition
````

### 9.5 Common String Methods
A **method** is a function that "belongs to" an object and is called using dot notation: `object.method()`. Strings have many built-in methods:


````
s = "  Hello World  "

print(s.upper())            # HELLO WORLD
print(s.lower())              # hello world
print(s.strip())                # 'Hello World' - removes leading/trailing whitespace
print(s.strip().replace("World", "Python"))   # 'Hello Python'
print(s.strip().split(" "))       # ['Hello', 'World'] - splits into a list

print("Hello".startswith("He"))     # True
print("Hello".endswith("lo"))         # True
print("Hello".find("l"))                # 2 - index of FIRST occurrence
print("Hello".count("l"))                 # 2 - how many times 'l' appears
print(len("Hello"))                         # 5 - length of the string

print(",".join(["a", "b", "c"]))               # 'a,b,c' - joins a list into a string
````

### 9.6 String Formatting
There are three ways to insert values into a string. **f-strings** (Python 3.6+) are the modern, preferred way.


````
name = "Ravi"
score = 88.5

# Method 1: % formatting (old style)
print("Name: %s, Score: %.1f" % (name, score))

# Method 2: .format()
print("Name: {}, Score: {}".format(name, score))

# Method 3: f-strings (BEST - use this in almost all cases)
print(f"Name: {name}, Score: {score:.1f}")     # :.1f means "1 decimal place"
````

### 9.7 Escape Sequences
Special character combinations starting with `\` that represent characters you can't type directly into a string.

| Sequence | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Tab |
| `\\` | A literal backslash |
| `\'` | A literal single quote |
| `\"` | A literal double quote |


````
print("Line1\nLine2")             # newline
print("Name:\tAge:")                # tab
print("She said \"hello\"")          # escaped double quotes
print('It\'s Python')                  # escaped single quote
print("C:\\Users\\Student")            # escaped backslashes (e.g. file paths)
````

## 10. Conditional Logic

### Definition
A **conditional statement** allows a program to make decisions — to execute different blocks of code depending on whether a condition (a boolean expression) is `True` or `False`. This is the foundation of all "logic" in programming.

### Syntax
```python
if condition:
    # runs if condition is True
elif another_condition:
    # runs if the first was False AND this one is True
else:
    # runs if ALL above conditions were False
```
`elif` (short for "else if") and `else` are both **optional**. You can have zero or many `elif` blocks, and at most one `else`.


````
marks = 82

if marks >= 90:
    print("Grade: A+")
elif marks >= 75:
    print("Grade: A")
elif marks >= 50:
    print("Grade: B")
else:
    print("Grade: F")

# Only ONE block runs - the first one whose condition is True.
# Once a match is found, Python skips the rest entirely.
````

### 10.1 Nested Conditionals
An `if` statement can contain another `if` statement inside it — this is called **nesting**, and is used when a decision depends on multiple, layered conditions.


````
age = 20
has_ticket = True

if age >= 18:
    if has_ticket:
        print("Entry allowed")
    else:
        print("Entry denied - no ticket")
else:
    print("Entry denied - underage")
````

### 10.2 Combining Conditions with Logical Operators
Often cleaner than nesting — use `and`/`or`/`not` (from Section 8.3) to combine multiple checks into a single condition.


````
age = 20
has_ticket = True

# Same logic as the nested example above, written flatter using 'and'
if age >= 18 and has_ticket:
    print("Entry allowed")
else:
    print("Entry denied")
````

### 10.3 The Ternary (Conditional) Expression
A compact, one-line way to assign a value based on a condition:
```python
value = value_if_true if condition else value_if_false
```


````
marks = 45
result = "Pass" if marks >= 40 else "Fail"
print(result)
````

## 11. Loops — In Depth

### Definition
A **loop** repeats a block of code multiple times, either a fixed number of times, or until a condition becomes `False`. Loops are essential for avoiding repetitive code and for processing collections of data.

### 11.1 The `for` Loop
Used to **iterate** (go through, one at a time) over a sequence — a string, a list, a `range()`, etc. `for` is used when you know (or can determine) what you're looping *over*.


````
# Iterating over a string
for letter in "ABC":
    print(letter)

print("---")

# Iterating over a list
colors = ["red", "green", "blue"]
for color in colors:
    print(color)
````

### 11.2 `range()`
`range(start, stop, step)` generates a sequence of numbers, commonly used to loop a specific number of times.
- `range(stop)` → `0` to `stop - 1`
- `range(start, stop)` → `start` to `stop - 1`
- `range(start, stop, step)` → `start` to `stop - 1`, incrementing by `step`

Just like slicing, `stop` is always **exclusive**.


````
for i in range(5):           # 0,1,2,3,4
    print(i, end=" ")
print()

for i in range(2, 8):          # 2,3,4,5,6,7
    print(i, end=" ")
print()

for i in range(0, 20, 5):        # 0,5,10,15
    print(i, end=" ")
print()

for i in range(10, 0, -2):         # counting DOWN: 10,8,6,4,2
    print(i, end=" ")
````

### 11.3 The `while` Loop
Repeats a block of code **as long as a condition remains `True`**. Used when you *don't* know in advance exactly how many times you need to loop — you loop until some condition changes.

⚠️ You are responsible for making sure the condition eventually becomes `False` — otherwise you get an **infinite loop**.


````
count = 0
while count < 5:
    print("Count is:", count)
    count += 1     # CRITICAL - without this, the condition never changes, and the loop never ends

print("Loop finished")
````

### 11.4 `break`, `continue`, and `pass`
- **`break`** — immediately exits the loop entirely, skipping any remaining iterations.
- **`continue`** — skips the *rest of the current iteration* only, and moves on to the next one.
- **`pass`** — does absolutely nothing; a placeholder used when Python's syntax requires a statement, but you don't want any action to happen yet.


````
# break
for i in range(1, 10):
    if i == 5:
        break             # stop the loop completely once i reaches 5
    print(i)

print("---")

# continue
for i in range(1, 10):
    if i % 2 == 0:
        continue           # skip even numbers, but keep looping
    print(i)

print("---")

# pass - useful as a placeholder while you're still planning your code
for i in range(3):
    if i == 1:
        pass    # "I'll handle this case later" - does nothing for now
    print(i)
````

### 11.5 Nested Loops
A loop inside another loop. The inner loop completes **all** of its iterations for **every single** iteration of the outer loop. This is the key technique behind pattern printing and grid/table-style logic.


````
for row in range(1, 4):            # outer loop: 3 rows
    for col in range(1, 4):           # inner loop: runs fully, 3 times, for EACH row
        print(f"({row},{col})", end=" ")
    print()   # move to a new line after each row completes
````

### 11.6 The `else` Clause on Loops
An often-unknown Python feature: both `for` and `while` loops can have an `else` block, which runs **only if the loop completed normally, without hitting a `break`**.


````
# Example: searching for a number, using loop-else to detect "not found"
numbers = [2, 4, 6, 8]
target = 5

for n in numbers:
    if n == target:
        print("Found it!")
        break
else:
    print("Not found - loop completed without a break")
````

## 12. Logic Building — Thinking Like a Programmer

This section isn't a new syntax topic — it's about **combining everything learned so far** (conditionals + loops + operators) to solve problems methodically. This is the actual skill of "programming," beyond just knowing syntax.

### A Simple Method for Any Logic-Building Problem
1. **Understand the problem** — restate it in your own words. What is the input? What is the expected output?
2. **Work it out by hand first** — solve a small example on paper before writing any code.
3. **Break it into steps** (pseudocode) — write the logic in plain English/bullet points.
4. **Translate each step into Python**, one piece at a time.
5. **Test with multiple inputs**, including edge cases (0, negative numbers, empty input).

### Worked Example: "Is a number a palindrome?" (e.g., 121 is, 123 is not)

**Step 1 — Understand:** A number reads the same forwards and backwards.

**Step 2 — By hand:** `121` reversed is `121` → same → palindrome. `123` reversed is `321` → different → not.

**Step 3 — Pseudocode:**
```
1. Store the original number
2. Reverse the number (using math, not string tricks, to practice loop logic)
3. Compare the reversed number to the original
4. If equal -> palindrome, else -> not
```

**Step 4 — Code:**


````
number = 121
original = number
reversed_number = 0

while number > 0:
    last_digit = number % 10             # extract the last digit
    reversed_number = reversed_number * 10 + last_digit   # build the reversed number
    number = number // 10                  # remove the last digit

if original == reversed_number:
    print(f"{original} is a palindrome")
else:
    print(f"{original} is not a palindrome")
````

**Step 5 — Test with more inputs:**

````
def check_palindrome(n):
    original = n
    reversed_number = 0
    while n > 0:
        last_digit = n % 10
        reversed_number = reversed_number * 10 + last_digit
        n = n // 10
    return original == reversed_number

for test_value in [121, 123, 1, 1221, 100]:
    print(test_value, "->", check_palindrome(test_value))
# (We formally define functions like this in Section 13 - used here as a preview.)
````

### Worked Example: Pattern Printing (Nested Loop Logic)

Pattern problems are one of the best ways to practice nested-loop thinking. The trick is always to ask: **"What controls the rows? What controls the columns? What decides what gets printed in each cell?"**


````
# Pattern 1: Right triangle of stars
rows = 5
for i in range(1, rows + 1):
    print("*" * i)
````

````
# Pattern 2: Number triangle
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

````
# Pattern 3: Inverted triangle of stars
rows = 5
for i in range(rows, 0, -1):
    print("*" * i)
````

````
# Pattern 4: Pyramid (using spaces + stars)
rows = 5
for i in range(1, rows + 1):
    spaces = " " * (rows - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
````

## 13. Functions

### Definition
A **function** is a named, reusable block of code designed to perform a specific task. Functions let you write logic once and run it as many times as you need, with different inputs — this avoids repetition and makes programs easier to organize, test, and debug.

### Anatomy of a Function
```python
def function_name(parameters):
    """Optional docstring describing what the function does."""
    # function body
    return value   # optional
```
- `def` — the keyword that begins a function definition.
- **Parameters** — named placeholders for the values a function expects to receive (defined at the function's creation).
- **Arguments** — the actual values you pass in when *calling* the function.
- **`return`** — sends a value back to wherever the function was called from, and immediately ends the function's execution. A function with no `return` statement implicitly returns `None`.


````
def greet(name):
    """Takes a name and returns a greeting string."""
    return f"Hello, {name}!"

message = greet("Aditi")     # "Aditi" is the ARGUMENT, matched to the 'name' PARAMETER
print(message)

print(greet("Ravi"))            # calling the SAME function again, with a different argument
````

### 13.1 Default Parameters
A parameter can have a **default value**, used automatically if the caller doesn't provide that argument.


````
def power(base, exponent=2):     # exponent defaults to 2 if not given
    return base ** exponent

print(power(5))          # uses default: 5**2 = 25
print(power(5, 3))       # overrides default: 5**3 = 125
````

### 13.2 Positional vs Keyword Arguments
- **Positional arguments** are matched to parameters by their **order**.
- **Keyword arguments** are matched by explicitly naming the parameter, so order no longer matters.


````
def describe_student(name, age, course):
    print(f"{name}, age {age}, studying {course}")

describe_student("Kabir", 20, "Python")                          # positional
describe_student(name="Kabir", course="Python", age=20)            # keyword - order doesn't matter
describe_student("Kabir", course="Python", age=20)                   # mixed - positional MUST come first
````

### 13.3 `*args` and `**kwargs`
Used when a function needs to accept a **variable, unknown number** of arguments.
- `*args` collects extra **positional** arguments into a tuple.
- `**kwargs` collects extra **keyword** arguments into a dictionary.


````
def add_all(*numbers):          # numbers becomes a tuple of whatever is passed in
    total = 0
    for n in numbers:
        total += n
    return total

print(add_all(1, 2))
print(add_all(1, 2, 3, 4, 5))       # works with ANY number of arguments


def print_profile(**details):        # details becomes a dictionary
    for key, value in details.items():
        print(f"{key}: {value}")

print_profile(name="Ravi", age=22, city="Mumbai")
````

### 13.4 Variable Scope
**Scope** determines where in your code a variable can be accessed.
- A **local variable** is created inside a function and only exists (is only accessible) inside that function.
- A **global variable** is created outside any function and can be *read* from anywhere, but to *modify* it from inside a function, you must explicitly declare it with the `global` keyword.


````
counter = 0    # GLOBAL variable

def increment_wrong():
    counter = counter + 1   # this creates a NEW local variable, causes an error - Python doesn't know
                              # if you meant to read the global one before it's assigned locally
try:
    increment_wrong()
except UnboundLocalError as e:
    print("Error:", e)

def increment_correct():
    global counter            # explicitly say: I mean the GLOBAL counter, not a new local one
    counter = counter + 1

increment_correct()
increment_correct()
print("Counter is now:", counter)
````

## 14. Recursion (Introductory)

### Definition
**Recursion** is when a function calls **itself** in order to solve a problem, by breaking it down into smaller versions of the same problem. Every correct recursive function needs:
1. A **base case** — the simplest version of the problem, which stops the recursion (without this, you get infinite recursion and a crash).
2. A **recursive case** — where the function calls itself with a smaller/simpler input, moving toward the base case.

### Example: Factorial
Mathematically: `5! = 5 × 4 × 3 × 2 × 1`, and also `5! = 5 × 4!`, `4! = 4 × 3!`, and so on, until `1! = 1` (the base case).


````
def factorial(n):
    if n == 0 or n == 1:      # BASE CASE - stops the recursion
        return 1
    else:                        # RECURSIVE CASE - problem broken into a smaller version of itself
        return n * factorial(n - 1)

print(factorial(5))    # 5 * 4 * 3 * 2 * 1 = 120
print(factorial(0))     # base case directly
print(factorial(7))
````

**Tracing through `factorial(4)` step by step:**
```
factorial(4)
= 4 * factorial(3)
= 4 * (3 * factorial(2))
= 4 * (3 * (2 * factorial(1)))
= 4 * (3 * (2 * 1))                <- base case reached, returns 1
= 4 * (3 * 2)
= 4 * 6
= 24
```

### Example: Sum of first N natural numbers


````
def sum_upto(n):
    if n == 0:              # base case
        return 0
    return n + sum_upto(n - 1)    # recursive case

print(sum_upto(5))    # 5+4+3+2+1+0 = 15
````

## 15. Data Structures — List, Tuple, Set, Dictionary

### Definition
A **data structure** is a way of organizing and storing multiple values together, so they can be accessed and worked with efficiently. Python has four built-in collection types, each with different rules:

| Structure | Ordered? | Mutable? | Allows Duplicates? | Written As |
|---|---|---|---|---|
| **List** | Yes | Yes | Yes | `[1, 2, 3]` |
| **Tuple** | Yes | No | Yes | `(1, 2, 3)` |
| **Set** | No | Yes | No | `{1, 2, 3}` |
| **Dictionary** | Yes (3.7+) | Yes | Keys must be unique | `{"key": "value"}` |

### 15.1 List
An **ordered, mutable** collection — the most commonly used data structure in Python.


````
fruits = ["apple", "banana", "cherry"]

# Accessing - same indexing/slicing rules as strings
print(fruits[0])
print(fruits[-1])
print(fruits[0:2])

# Modifying - lists are MUTABLE
fruits.append("mango")           # add to the end
fruits.insert(1, "orange")         # insert at a specific index
fruits.remove("banana")              # remove by value
popped = fruits.pop()                  # removes AND returns the last item
fruits[0] = "green apple"                # update by index

print(fruits)
print("Popped item was:", popped)
print("Length:", len(fruits))

# Other common operations
numbers = [5, 3, 8, 1, 9]
numbers.sort()                # sorts in place (modifies the original list)
print(numbers)
numbers.reverse()
print(numbers)
print(sum(numbers), min(numbers), max(numbers))
````

### 15.2 Tuple
An **ordered, immutable** collection. Use it when you want to guarantee the data cannot be changed after creation (e.g., fixed coordinates, RGB color values).


````
point = (10, 20)
print(point[0], point[1])

try:
    point[0] = 99
except TypeError as e:
    print("Error:", e)   # tuples cannot be modified after creation

# Tuple unpacking - a very common, convenient Python pattern
x, y = point
print(f"x={x}, y={y}")
````

### 15.3 Set
An **unordered collection of unique values** — automatically removes duplicates, and is optimized for fast membership checks (`in`).


````
numbers = [1, 2, 2, 3, 3, 3, 4]
unique_numbers = set(numbers)
print(unique_numbers)             # duplicates automatically removed

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print("Union:", set_a | set_b)             # all elements from both
print("Intersection:", set_a & set_b)        # only elements in BOTH
print("Difference:", set_a - set_b)            # in set_a but NOT in set_b

set_a.add(10)
set_a.discard(1)
print(set_a)
````

### 15.4 Dictionary
An **ordered collection of key-value pairs**. Each **key** must be unique and is used to look up its associated **value** — like a real dictionary, where you look up a word (key) to find its meaning (value).


````
student = {
    "name": "Ravi",
    "age": 21,
    "courses": ["Python", "Django"]
}

# Accessing values
print(student["name"])
print(student.get("email", "Not Provided"))     # .get() is SAFE - won't crash if key doesn't exist

# Adding / Updating
student["email"] = "ravi@example.com"     # adds a new key
student["age"] = 22                         # updates an existing key

# Removing
del student["age"]

# Iterating
for key, value in student.items():
    print(key, "->", value)

print("keys:", list(student.keys()))
print("values:", list(student.values()))
print("age" in student)   # checks KEYS by default when using 'in' on a dict
````

## 16. Comprehensions

A **comprehension** is a concise, single-line way to build a list, dictionary, or set from an existing sequence — combining a `for` loop (and optionally an `if` filter) into one expression. This is considered idiomatic ("Pythonic") code.

### General Form
```python
[expression for item in sequence if condition]
```
This is equivalent to writing:
```python
result = []
for item in sequence:
    if condition:
        result.append(expression)
```


````
# List comprehension - build a list of squares
squares = [x ** 2 for x in range(1, 6)]
print(squares)

# The traditional loop equivalent, for comparison
squares_traditional = []
for x in range(1, 6):
    squares_traditional.append(x ** 2)
print(squares_traditional)   # identical result, more lines of code
````

````
# With a condition (filter)
evens = [x for x in range(1, 20) if x % 2 == 0]
print(evens)

# Dictionary comprehension
names = ["Amit", "Priya", "Zoya"]
name_lengths = {name: len(name) for name in names}
print(name_lengths)

# Set comprehension
unique_lengths = {len(name) for name in names}
print(unique_lengths)
````
